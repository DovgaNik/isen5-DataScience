```python
model = "llama3.2:1b"
prompt = "What is Charles de Gaulle, is it a ship ?"
CSV_FILE = "student_data.csv"
```

```python
def generateOne(question: str):
    response = ollama.chat(
        model=model,
        messages=[{"role": "user", "content": question}])
    return response["message"]["content"]
```

```python
generateOne("Write HelloWorld in assembly")
```

```python
print("=== DATA SCIENCE: DATA COLLECTION ===")
print("Topic: Study habits and quiz performance")
print("Use fictional observations for this lab.")
number_of_students = int(input("Number of records to collect: "))
dataset = []
for i in range(number_of_students):
    print(f"Participant {i + 1}")
    study_hours = float(
    input("Study hours per day: "))
    sleep_hours = float(
    input("Sleep hours per night: "))
    quiz_score = float(input("Quiz score (0-100): "))

record = {
"participant_id": i + 1,
"study_hours": study_hours,
"sleep_hours": sleep_hours,
"quiz_score": quiz_score
}
dataset.append(record)
```

```python
with open(CSV_FILE, "w", newline="", encoding="utf-8") as file:
    columns = [
    "participant_id",
    "study_hours",
    "sleep_hours",
    "quiz_score"
    ]
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()
    writer.writerows(dataset)
    print(f"Dataset saved to {CSV_FILE}")
```
